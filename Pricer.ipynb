# Structured Products Refactor Demo

This notebook preserves the copied legacy Snowball / FCN notebook and demonstrates the new autocall framework.

```python
from structured_products import (
    EngineConfig,
    MarketData,
    legacy_price,
    make_butterfly_autocall,
    make_classic_autocall,
    make_dividend_autocall,
    make_stepdown_autocall,
    make_wide_autocall,
    mc_greeks,
    price,
    support_matrix,
)
```

```python
market = MarketData(rate=0.03, dividend_yield=0.0, volatility=0.2)
mc_config = EngineConfig(n_paths=12000, seed=7)
pde_config = EngineConfig(pde_spot_steps=300)
support_matrix()
```

## Legacy product example

```python
legacy_snowball = legacy_price(
    method='PDE',
    Type=2,
    margin_rate=1.0,
    S=1.0,
    K=1.0,
    T=1.0,
    R=0.18,
    N=1.0,
    Ll=0.8,
    Lh=1.0,
    r=0.03,
    q=0.0,
    sigma=0.2,
    size=10000,
    obs_start_day=21,
    obs_in=0,
    day_counter=252,
    _freq_num=21,
    KI_freq='daily',
    KO_freq='monthly',
    RFlag=0,
)
legacy_snowball
```

## Classic autocall

```python
classic = make_classic_autocall()
{
    'mc': price(classic, market, method='MC', engine_config=mc_config),
    'pde': price(classic, market, method='PDE', engine_config=pde_config),
    'greeks': mc_greeks(classic, market, mc_config),
}
```

## Wide autocall

```python
wide = make_wide_autocall()
{
    'mc': price(wide, market, method='MC', engine_config=mc_config),
    'pde': price(wide, market, method='PDE', engine_config=pde_config),
}
```

## Dividend autocall

```python
dividend = make_dividend_autocall()
{
    'mc': price(dividend, market, method='MC', engine_config=mc_config),
    'pde': price(dividend, market, method='PDE', engine_config=pde_config),
}
```

## Butterfly autocall

```python
butterfly = make_butterfly_autocall(front_coupon=0.22, back_coupon=0.10)
{
    'mc': price(butterfly, market, method='MC', engine_config=mc_config),
    'pde': price(butterfly, market, method='PDE', engine_config=pde_config),
}
```

## Step-down autocall

```python
stepdown = make_stepdown_autocall(first_knock_out_ratio=1.03, step_down_ratio=0.01)
{
    'mc': price(stepdown, market, method='MC', engine_config=mc_config),
    'note': 'PDE is intentionally disabled for step-down in this version',
}
```
